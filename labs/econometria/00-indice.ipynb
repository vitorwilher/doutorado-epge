## Como rodar

Cada notebook abre no **Google Colab** com um clique, roda no navegador e não exige
instalação nem conta paga. O runtime é o de **R** — os pacotes `ggplot2` e `dplyr` já
vêm instalados, e a primeira célula de cada notebook instala o que faltar.

> **Dica:** rode as células em ordem (`Shift+Enter`). As simulações mais pesadas
> (Aula 6, seções 4 e 5) levam cerca de 30 segundos cada.

## Um aviso sobre os dados

Os exemplos de *TestScore* × *STR* usam dados **simulados e calibrados** para se
parecerem com os de Stock & Watson (2020) — não são os dados originais da Califórnia.
Isso é deliberado: numa simulação **conhecemos** os parâmetros verdadeiros, e é por
isso que dá para falar em viés com precisão, em vez de por analogia.

## Um fio condutor

Se houver uma única ideia para levar destes notebooks, é a diferença entre
**ruído** e **viés**:

- o **ruído** amostral encolhe quando a amostra cresce — é o que os erros-padrão e os
  intervalos de confiança medem;
- o **viés** de identificação **não encolhe** (Aula 5, seção 3). Mais dados apenas
  estimam o número errado com mais precisão.

Quase tudo o que o curso faz depois disso é uma estratégia para atacar a segunda coisa.
